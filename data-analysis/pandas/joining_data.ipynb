```python
print(wards_census.columns)
```

```python
### You may have noticed that the merged table has columns with suffices of underscore c or y.
### This is because both the wards and census tables contained address and zip columns.
### To avoid multiple columns with the same name, they are aoutomatically given a suffix by the merge method.
```

```python
wards_census = wards.merge(census, on='ward', suffixes=('_ward','_cen'))
print(wards_census.head())
print(wards_census.shape)
```

```python
### We can use the suffix argument of the merge method control this behavior.
### We provide a tuple where all of the overlapping columns in the left table are given the suffix '_ward',
### and those of the right table will be given the suffix '_cen'.
### This makes it easier for us to tell the difference between the columns.
```

```python
## One-to-one
```

```python
### In a one-to-one relationship, every row in the left table is related to one and only one row in the right table. 
```

```python
### Recall the relationship between the wards table and the census table.
### Every row in the ward table is related to only one row in the census table, so there is only one row for ward 3 in each table.
### Practically speaking, it only makes sense that there is one row of population information for each ward.
### It wouldn't make sense if the census table contained multiple population values in 2000 for the third ward.
```

```python
import pandas as pd
import numpy as np

sequel_data = {
    "id":    [19995, 862,  863,   10193, 597,   24428, 675,  767,   121,  122,  120,  121,  10193],
    "title": ["Avatar", "Toy Story", "Toy Story 2", "Toy Story 3", "Titanic", "The Avengers",
              "Harry Potter 1", "Harry Potter 2", "Lord of the Rings 1", "Lord of the Rings 2",
              "Lord of the Rings 3", "Lord of the Rings 2", "Toy Story 3"],
    "sequel": [np.nan, 863, 10193, np.nan, np.nan, np.nan, 767, np.nan, 122, 120, np.nan, 120, np.nan]
}

sequel = pd.DataFrame(sequel_data)
sequel.to_csv("sequel.csv", index=False)
sequel = pd.read_csv("sequel.csv")

print(sequel.head())
print(sequel.shape)
```

```python
### The table shown here is called sequels and has three columns. It contains a column for movie id, title and sequel.
### The sequel number refers to the movie id that is a sequel to the original movie.
```

```python
### For example, in the secon row the movie is titled Toy Story, and has an id equal to 862. The sequel number of this row is 863.
### This is the movie id for Toy Story 2, the sequel to Toy Story.
```

```python
from IPython.display import Image
Image("merging-table-itself.png")
```

```python
### If we would like to see a table with the movies and the corresponding sequel movie in one row of the table, we will need to merge the table to itself. 
### In the left table, the sequel ID for Toy Story of 863 is matched with 863 in the ID column of the right table.
### Similarly, Toy Story 2 of the left table is matched with Toy Story 3 in the right table.
```

```python
### To complete this merge, we set the sequels table as input to the merge method for both the left and right tables.
### We can think of it as merging two copies of the same table.
### All of the aspects we have reviewed regarding merging two tables still apply here. Therefore, we can merge the tables on different columns.
### We'll use the 'left_on' and 'right_on' attributes to match rows where the sequel's id matches the original movie's id.
### Finally, setting the suffixes argument in the merge method allows us to identify which column describe the original movie and which describe the sequel.
```

```python
original_sequel = sequel.merge(sequel, left_on='sequel', right_on='id',
                                 suffixes=('_org', '_seq'))
print(original_sequel.head())
```

```python
### When we look at the results of the merge, the 'title_or' and 'title_seq' list the original and sequel movies, respectively.
### Here we listed the original movie and its sequel in one row.
```

```python
## Continue format results
```

```python
print(original_sequel[['title_org', 'title_seq']].head())
```

```python
### Now that we have our result table from the merge, we could selecet only the 'title_org' and 'title_seq' columns, and we can see that we've successfully linked each movie to its sequel.
```

```python
import numpy as np
import pandas as pd

songs_data = {
    "tid": [1, 2, 3, 4 , 5],
    "name": ["For Those About to Rock We Salute You", "Balls To The Wall", "Fast As a Shark", "Restless and Wild", "Primcess of the Dawn"],
    "aid": [1, 2, 3, 3, 3],
    "mtid": [1, 2, 2, 2, 2],
    "gid": [1, 1, 1, 1, 1],
    "composer": ["Angus Young", np.nan, "F. Baltes, S. Kaufmann", "F. Baltes, R. A. Smith", "Deaffy & R.A. Smith"],
    "u_price": [0.99, 0.99, 0.99, 0.99, 0.99]   
}

songs = pd.DataFrame(songs_data)
songs.to_csv("songs.csv", index=False)
songs = pd.read_csv("songs.csv")
print(songs.head())
```

### **Topic 3.1: Semi-Joins**
A semi-join filters the left table down to only those observations that have a match in the right table. Unlike an inner join, it only returns columns from the left table, and it avoids duplicate rows.

```python
# Performing a semi-join
# Step 1: Merge the tables on key column 'gid' (genres)
genres_songs = songs.merge(musics, on='gid')
print('Inner join shape:', genres_songs.shape)

# Step 2: Use .isin() to filter the left table (songs)
semi_join = songs[songs['gid'].isin(musics.index)]
print('Semi-join shape:', semi_join.shape)
print(semi_join.head())
```

### **Topic 3.2: Anti-Joins**
An anti-join filters the left table to only those observations that **do not** have a match in the right table. It only returns columns from the left table.

```python
# Performing an anti-join
# Step 1: Left join with indicator=True
songs_genres_indicator = songs.merge(musics, on='gid', how='left', indicator=True)
print(songs_genres_indicator.head())

# Step 2: Filter where _merge is 'left_only'
anti_join = songs_genres_indicator[songs_genres_indicator['_merge'] == 'left_only']

# Step 3: Keep only columns from the left table
left_cols = songs.columns
anti_join = anti_join[left_cols]
print('Anti-join shape:', anti_join.shape)
print(anti_join.head())
```

### **Topic 3.3: Concatenating DataFrames vertically and horizontally**
We use `pd.concat()` to append rows (vertical) or columns (horizontal) from multiple DataFrames.

```python
# Create sample DataFrames for concatenation
songs_pop = songs.copy()
songs_rock = songs.copy()

# Vertical concatenation (stacking rows)
stacked_songs = pd.concat([songs_pop, songs_rock], ignore_index=True)
print('Stacked shape:', stacked_songs.shape)

# Horizontal concatenation (aligning columns on index)
side_by_side = pd.concat([songs_pop, songs_rock], axis=1)
print('Side-by-side shape:', side_by_side.shape)
```

### **Topic 3.4: Verifying Integrity**
Pandas allows us to check that relations are maintained during merges and concatenations, raising errors if validation checks fail.
- In `.merge()`, use the `validate` argument: `'one_to_one'`, `'one_to_many'`, `'many_to_one'`, `'many_to_many'`.
- In `pd.concat()`, use the `verify_integrity=True` argument to prevent duplicate indices.

```python
try:
    # Validate a one-to-one relationship when merging (will fail if many-to-one)
    songs.merge(musics, on='gid', validate='one_to_one')
except ValueError as e:
    print('Validation Failed as expected:', e)

try:
    # Verify integrity on concat (will fail if index values are duplicated)
    pd.concat([songs, songs], verify_integrity=True)
except ValueError as e:
    print('Concat integrity check failed as expected:', e)
```

### **Topic 4.1: pd.merge_ordered()**
Useful for time-series data or sorted fields. It performs an outer join by default and sorts the results. It also supports forward-filling missing values.

```python
# Sample ordered time series data
df_a = pd.DataFrame({'date': pd.to_datetime(['2026-06-01', '2026-06-03']), 'val_a': [10, 20]})
df_b = pd.DataFrame({'date': pd.to_datetime(['2026-06-02', '2026-06-04']), 'val_b': [15, 25]})

# Merge ordered
ordered_merge = pd.merge_ordered(df_a, df_b, on='date')
print('Ordered Merge:\n', ordered_merge)

# Forward fill missing values
ffill_merge = pd.merge_ordered(df_a, df_b, on='date', fill_method='ffill')
print('With forward fill:\n', ffill_merge)
```

### **Topic 4.2: pd.merge_asof()**
Similar to `pd.merge_ordered()`, but it matches on the nearest key rather than an exact key. Direction can be 'backward' (default), 'forward', or 'nearest'.

```python
# Sample trade logs and quotes
trades = pd.DataFrame({
    'time': pd.to_datetime(['2026-06-11 12:00:01', '2026-06-11 12:00:04']),
    'price': [100.5, 101.2]
})
quotes = pd.DataFrame({
    'time': pd.to_datetime(['2026-06-11 12:00:00', '2026-06-11 12:00:03']),
    'bid': [100.4, 101.1]
})

# Merge asof: finds the last quote prior to or equal to the trade time
asof_merge = pd.merge_asof(trades, quotes, on='time')
print('Asof Merge (Backward):\n', asof_merge)
```

### **Topic 4.4: Reshaping Data with pd.melt()**
Converting a DataFrame from a wide format (multiple columns for variables) to a long format (variable-value pairs).

```python
# Create wide format DataFrame
wide_df = pd.DataFrame({
    'movie': ['Toy Story', 'Titanic'],
    'rating_imdb': [8.3, 7.9],
    'rating_rotten': [100, 89]
})
print('Wide DataFrame:\n', wide_df)

# Melt into long format
long_df = pd.melt(wide_df, id_vars=['movie'], 
                  value_vars=['rating_imdb', 'rating_rotten'],
                  var_name='source', value_name='score')
print('Melted (Long) DataFrame:\n', long_df)
```
