```python
import pandas as pd

dogs_df = {
    "dogs_id": [f"{str(i).zfill(3)}" for i in range (1,8)],
    "name": ["Bella", "Charlie", "Lucy", "Cooper", "Max", "Stella", "Bernie"],
    "breed": ["Labrador", "Poodle", "Chow Chow", "Schnauzer", "Labrador", "Chihuahua", "St. Bernard"],
    "color": ["Brown", "Black", "Brown", "Gray", "Black", "Tan", "White"],
    "height_cm": [56, 43, 46, 49, 59, 18, 77],
    "weight_kg": [25, 23, 22, 17, 29, 2, 74],
    "date_of_birth": ["2013-07-01", "2016-09-16", "2014-08-25", "2011-12-11", "2017-01-20", "2015-04-20", "2018-02-27"],
}
dogs = pd.DataFrame(dogs_df)
print(dogs)
print(dogs.head()) #  returns the first few rows (the “head” of the DataFrame).
print(dogs.info()) # shows information on each of the columns, such as the data type and number of missing values.
print(dogs.shape) # returns the number of rows and columns of the DataFrame.
print(dogs.describe()) # calculates a few summary statistics for each column.
print(dogs.to_numpy()) # values: A two-dimensional NumPy array, accessed via the .to_numpy() method.
print(dogs.columns) # columns: An index of column names, accessed via the .columns attribute.
print(dogs.index) # index: An index for the rows, accessed via the .index attribute.
```

```python
## The first thing you can do is change the order of the rows by sorting them
## So that the most interesting data is at the top of the DataFrame
## You can sort rows using the sort_values method, passing in a column name that you want to sort by.
## For example, when we apply sort_values on the weight_kg column of the dogs DataFrame, 
## we get the lightest dog ar the top, Stella the Chihuahua

dogs.sort_values("weight_kg")
```

```python
# Sorting and subsetting
```

```python
## Sorting in descending order
```

```python
### Setting the ascending argument to False will sort the data the other way around,
### from heaviest dog to lightest dog.

dogs.sort_values("weight_kg", ascending=False)
```

```python
# Sorting by multiple variables
```

```python
## We can sort by multiple variables by passing a list of column names to sort_values.
dogs.sort_values(["weight_kg", "height_cm"])
```

```python
## Now, Charlie, Lucy and Bella are ordered from shortest to tallest, even though they all weigh the same.

dogs.sort_values(["weight_kg", "height_cm"], ascending=[True, False])
## To change the direction values are sorted in, 
## pass a list to the ascending argument to specify wich direction sorting sould be done for each variable.
```

```python
# Methods

## Methods have four key components:

### Name: Identifies the operation, like sum()
### Target: The dot (.) specifies which data the method operates on. For example, amounts.sum() applies sum() to the amounts Series
### Parameters: Additional inputs go inside (). If none are needed, the parentheses stay empty
### Output: The method returns a result. For example, sum() returns the total
### Think of it like this: target.method(parameters) → output
```

```python
## Pivot tables ar another way of calculating grouped summary statistics.
## If you've ever used a spreadcheet, chances are you've used a pivot table.
```

```python
# Group by to pivot table
```

```python
dogs.groupby("color")["weight_kg"].mean()
```

```python
## If we set the margins argument to True, 
## the last row and last column of the pivot table contain the mean of all values in the column or row,
## not including the missing values that were filled in with Os.
```

```python
dogs.pivot_table(values="weight_kg", index="color", columns="breed", fill_value=0, margins=True)
```

```python
### For example, we have some new dog data to put into a DataFrame. Let's start with the first method to do this, creating a list of dictionaries.
### First, we'll create a new list using square brackets to hold our dictionaries. Then, we'll go thtough the first row of our data and put in in a dictionary.
```

```python
list_of_dicts = [
    {"name": "Ginger", "breed": "Dachshund", "height_cm": 22,
    "weight_kg": 10, "date_of_birth": "2019-03-14"},
    {"name": "Scout", "breed": "Dalmatian", "height_cm": 59,
    "weight_kg": 25, "date_of_birth": "2019-05-09"}
]
```

```python
### When using this method, we need to go through the data column by column. 
### Key -> column name
### Value -> list of column values

### Each key vill be a column name, and each value will be a list of the values in the column.
```

```python
dict_of_lists ={
    "name": ["Ginger", "Scout"],
}
```
