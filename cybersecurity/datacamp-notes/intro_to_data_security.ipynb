## **Chapter 1: Cybersecurity & Data Security Foundations**

Data security is anchored on three core pillars:
1.  **Confidentiality:** Restricting access to authorized users (Access lists, encryption).
2.  **Integrity:** Preventing unauthorized modification (Hashing, digital signatures).
3.  **Availability:** Ensuring reliable access to authorized users (Redundancy, backups).

### **1.1: Compliance Regulations**
- **GDPR (General Data Protection Regulation):** Protects EU citizens' data, mandates "Right to be Forgotten".
- **HIPAA (Health Insurance Portability and Accountability Act):** Safeguards protected health information (PHI).
- **CCPA (California Consumer Privacy Act):** Gives consumers rights over personal data.
- **PCI-DSS (Payment Card Industry Data Security Standard):** Audits credit card storage security.

### **1.2: Data Lifecycle**
Data undergoes six phases: **Creation** $\to$ **Storage** $\to$ **Use** $\to$ **Share** $\to$ **Archive** $\to$ **Destruction**.

```python
# Simulating an audit trail logging function during data creation
import datetime

def log_data_event(event_type, user_id, details):
    log_entry = {
        'timestamp': datetime.datetime.now().isoformat(),
        'event': event_type,
        'user': user_id,
        'details': details
    }
    print(f'[AUDIT TRAIL] {log_entry}')

log_data_event('DATA_CREATION', 'system_admin', 'Initialized customer database schema.')
```

## **Chapter 2: Data Encryption, Key Management, and Hashing**

### **2.1: Symmetric Encryption (AES)**
Both encryption and decryption utilize the same key: $c = E_k(m), \, m = D_k(c)$.

### **2.2: Asymmetric Encryption (RSA & Diffie-Hellman)**
Uses public and private key pairs. RSA relies on prime factorization: $c \equiv m^e \pmod n, \, m \equiv c^d \pmod n$.

### **2.3: Password Hashing & Salts**
Hashing is one-way. To prevent dictionary/rainbow table attacks, we append a random string called a **Salt** to the password before hashing.

```python
import hashlib
import secrets

password = 'SecurePassword123'

# Generate a cryptographically secure random salt
salt = secrets.token_hex(16)
print('Generated Salt:', salt)

# Compute SHA-256 Hash with Salt
salted_password = (password + salt).encode()
pwd_hash = hashlib.sha256(salted_password).hexdigest()
print('Salted Password Hash:', pwd_hash)

# Verification process
entered_password = 'SecurePassword123'
verification_hash = hashlib.sha256((entered_password + salt).encode()).hexdigest()
print('Password Verification Success:', pwd_hash == verification_hash)
```

### **2.4: Key Vaults & Management**
Key management is the bottleneck of encryption safety:
- **KMS (Key Management Service):** Securely stores, deletes, and rotates cryptographic keys.
- **HSM (Hardware Security Module):** Physical hardware devices that generate and safeguard keys.
- **Key Rotation:** Periodically replacing active keys to limit exposure in case of compromise.

## **Chapter 3: Access Control Models & Data Obfuscation**

### **3.1: Access Control Frameworks**
- **RBAC (Role-Based Access Control):** Permissions linked to user roles (e.g. admin, writer).
- **ABAC (Attribute-Based Access Control):** Evaluated dynamically using multiple criteria:
  $$\text{Access Granted} = f(\text{User Attributes}, \text{Resource Attributes}, \text{Environment})$$
- **DAC (Discretionary Access Control):** Resource owners decide who has access.
- **MAC (Mandatory Access Control):** Central operating systems enforce access permissions based on security clearances (e.g. military top secret classifications).

### **3.2: Static vs. Dynamic Data Masking & Tokenization**
- **Static Data Masking (SDM):** Permanently anonymizes dataset copies (ideal for non-production environments).
- **Dynamic Data Masking (DDM):** Redacts query outputs in real-time.
- **Tokenization:** Replaces credit cards or sensitive fields with randomized reference strings (tokens).

```python
# Simulating dynamic masking & ABAC rules
user_role = 'customer_support'
credit_card_column = '4321-8765-9012-3456'

def query_credit_card(role, card):
    if role == 'billing_admin':
        return card  # No masking for billing admins
    else:
        # Apply dynamic masking
        return 'XXXX-XXXX-XXXX-' + card[-4:]

print('Query Output for billing_admin:', query_credit_card('billing_admin', credit_card_column))
print('Query Output for customer_support:', query_credit_card('customer_support', credit_card_column))
```

## **Chapter 4: Incident Response, SQL Injection, and SIEM**

### **4.1: SQL Injection (SQLi) Protection**
SQLi occurs when input is concatenated directly into SQL queries. We prevent this using parameterized queries (prepared statements).

```python
# Simulated SQLi Prevention
unsafe_user_input = "' OR '1'='1"

# Unsafe concatenation (SQL Injection vulnerability)
vulnerable_query = f"SELECT * FROM users WHERE username = '{unsafe_user_input}'"
print('Vulnerable Query String:\n', vulnerable_query)

# Safe Parameterization
safe_query = "SELECT * FROM users WHERE username = ?"
print('\nSafe query parameterized template:', safe_query)
print('Input passed separately as parameters tuple:', (unsafe_user_input,))
```

### **4.2: Incident Monitoring & Disaster Recovery**
- **SIEM (Security Information & Event Management):** Aggregates logs, parses them, and fires automated alerts.
- **Disaster Recovery Metrics:**
  - **RTO (Recovery Time Objective):** Maximum tolerable duration of system downtime.
  - **RPO (Recovery Point Objective):** Maximum tolerable age of data loss (e.g. backup frequency requirements).
